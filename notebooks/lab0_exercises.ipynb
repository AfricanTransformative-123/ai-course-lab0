```python
# Set seed for reproducibility
np.random.seed(42)

# Number of students
n_students = 200

# Create student dataset
data = {
    'student_id': range(1000, 1000 + n_students),
    'major': np.random.choice(
        ['Engineering', 'Economics', 'Computer Science', 'Business Administration'],
        n_students
    ),
    'year': np.random.choice([1, 2, 3, 4], n_students),
    'exam_score': np.random.normal(75, 10, n_students).clip(0, 100),
    'assignments_completed': np.random.randint(0, 11, n_students),
    'hours_studied': np.random.normal(15, 5, n_students).clip(1, 40)
}

# Create DataFrame
df = pd.DataFrame(data)

# Display first 5 rows
df.head()
```

```python
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Calculate average exam score for each major
average_scores = df.groupby('major')['exam_score'].mean()

# Display results
print(average_scores)
```

```python
# Plot distribution of exam scores
sns.histplot(df['exam_score'], kde=True)

# Add title
plt.title("Exam Score Distribution")

# Display plot
plt.show()
```

```python
# Create boxplot of exam scores by major
sns.boxplot(
    data=df,
    x='major',
    y='exam_score'
)

# Add title
plt.title("Scores by Major")

# Rotate x-axis labels
plt.xticks(rotation=10)

# Display plot
plt.show()
```

```python
# Create scatter plot of study hours vs exam scores
sns.scatterplot(
    data=df,
    x='hours_studied',
    y='exam_score',
    hue='major'
)

# Add title
plt.title("Hours Studied vs Exam Score")

# Display plot
plt.show()
```

```python
correlation = df[
    ['year', 'exam_score', 'assignments_completed', 'hours_studied']
].corr()

sns.heatmap(correlation, annot=True)

plt.title("Correlation Matrix")

plt.show()
```

```python
# Set seed for reproducibility
np.random.seed(42)

# Number of customers
n_customers = 500

# Generate customer data
ages = np.random.randint(18, 70, n_customers)
income = np.random.normal(50000, 20000, n_customers).clip(15000, 150000)
purchase_frequency = np.random.poisson(5, n_customers)
avg_purchase_value = np.random.normal(100, 30, n_customers).clip(10, 500)

# Create DataFrame
customers = pd.DataFrame({
    'age': ages,
    'income': income,
    'purchase_frequency': purchase_frequency,
    'avg_purchase_value': avg_purchase_value
})

# Display first 5 rows
customers.head()
```

```python
# Create scatter plot of income vs purchase value
sns.scatterplot(
    data=customers,
    x='income',
    y='avg_purchase_value'
)

# Add title
plt.title("Income vs Purchase Value")

# Display plot
plt.show()
```
